```r
source("../utilities/functions/genes-and-pathway-heatmaps-aesthetic.R")
set.seed(25)
```

```r
tissue <- "pbmc"
comp <- "PreTx_vs_EI"

pathways <- c(
  "INTERFERON GAMMA RESPONSE",
  "INTERFERON ALPHA RESPONSE",
  "INFLUENZA A",
  "EPSTEIN-BARR VIRUS INFECTION"
)

celltypes <- c(
  "CD8 T EM KLRF1- GZMB+",
  "CD8 T EM KLRF1+ GZMB+",
  "CD8 T EM GZMK- CD27+",
  "CD8 T EM GZMK+ CD27+",
  "CD8 T CM",
  "NK Adaptive",
  "CD56dim NK ISG+"
)

p <- plot_pathway_genes_heatmap(
  tissue = tissue,
  comp = comp,
  pathways = pathways,
  celltypes = rev(celltypes),
  celltype_level = "l3",
  input_dir = "../../data/figure-inputs/",
  timepoint_1 = "PreTx",
  timepoint_2 = "EI",
  title = "PBMC - PreTx vs EI",
  panel_heights = c(1.5, 4),
  cap = 2,
  x_text_size = 14,
  base_size = 12,
  plot_width = 17,
  plot_height = 5
)

p
ggsave("../../data/figure-outputs/supplementary-5/sup_5e_p1.pdf", plot = p, width = 17, height = 5, dpi = 200, device = cairo_pdf)
```

```r
# Load bone marrow data files
data <- read.csv('../../data/figure-inputs/l3-fgsea_all_results/fgsea_all_results_pbmc_PreTx_vs_EI.csv')
```

```r
plot_width = 6
plot_height = 5
plot_dpi = 200

# faceted heatmap for bone marrow
p <- create_pathway_heatmap(
  data = data,
  pathway_method = "manual",
  pathway_list = pathways,
  celltypes = rev(celltypes),
  max_label_length = 19,
  plot_width = plot_width,
  plot_height = plot_height,
  facet_ncol = 4
)

p
ggsave("../../data/figure-outputs/supplementary-5/sup_5e_p2.pdf", plot = p, width = plot_width, height = plot_height, dpi = plot_dpi, device = cairo_pdf)
```
