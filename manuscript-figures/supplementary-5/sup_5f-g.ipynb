## Sup 5F

```r
genes =  c("HIF1A", "LAMTOR1", "PGK1")
p <- plot_expression_violins(
  metadata_path = '../../data/figure-inputs/metadata/bmmc_sample_kit_metadata.csv',
  files_dir = '../../data/figure-inputs/bmmc_l3_lognorm_gexp_celltypes_per_samplekit',
  gene_list = genes,
  celltypes = c("CD8 T EM1", "CD8 T EM2", "CD8 T Tissue Res"),
  visits = c("PreTx", "EI", "ASCT90d", "ASCT1y", "ASCT2y", "Healthy"),
  paired_visits = c("PreTx", "EI", "ASCT90d", "ASCT1y", "ASCT2y"),
  unpaired_visit = "Healthy",
  plot_width = 6,
  plot_height = 7,
  base_size = 10, 
  step.increase = 0.06,
  subtitle = '',
  ylabs = 'Pseudobulk Gene Expression',
  pval_colors = c('maroon', 'seagreen', 'gray30', 'gray50'),
  line_size = 0.6
)
p

ggsave("../../data/figure-outputs/supplementary-5/sup_5f.pdf", plot = p, width = 6, height = 7, dpi = 200, device = cairo_pdf)
```
