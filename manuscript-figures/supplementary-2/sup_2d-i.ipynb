### Frequency Volcano Plot - H/L compared to healthy
Creates a volcano plot showing cell type frequency differences between two conditions. X-axis shows median difference, Y-axis shows -log10(FDR), with significant cell types labeled and colored by direction of change. Includes FDR correction and effect size thresholds for identifying significant changes.

```r
source('../utilities/functions/frequency-volcanos.R')
```

```r
plot_height <- 6
plot_width <- 6
plot_dpi <- 200
```

```r
# Main execution
df <- fread("../../data/figure-inputs/frequency_files/pbmc_l3_frequency.csv")
df <- df[!is.na(label.visitDetails) & label.visitDetails != "", ]

# Run analysis
deg_res <- test_wilcox_unpaired(df, g1 = "EI", g2 = "Healthy", freq_data = "cell_type_clr", label_col = "aifi_plot_l3")

options(repr.plot.width = plot_width, repr.plot.height = plot_height)
if (nrow(deg_res) > 0) {
  p <- plot_freq_volcano(df, deg_res, freq_data = "cell_type_clr", alpha = 0.05, effect_size = 0.05, title = "PBMC")
}

p
ggsave("../../data/figure-outputs/supplementary-2/sup_2d.pdf", plot = p, width = plot_width, height = plot_width, dpi = plot_dpi, device = cairo_pdf)
```
