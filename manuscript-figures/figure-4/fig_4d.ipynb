```r
tissue <- "bmmc"
comp <- "EI_vs_ASCT1y"

pathways <- c(
  "MAPK SIGNALING PATHWAY",
  "CYTOKINE SIGNALING IN IMMUNE SYSTEM",
  "TNF-ALPHA SIGNALING VIA NF-KB",
  "IL-17 SIGNALING PATHWAY"
)

celltypes <- c(
  "HSPC Stem", "HSPC Multi", "LMPP", "CLP",
  "Prog B Pre", "Prog B",
  "CMP Core", "CMP Gran",
  "MEP", "Pre Prog Ery",
  "Pre Mono Prolif",
  "Prog DC cDC", "Prog DC pDC"
)

p <- plot_pathway_genes_heatmap(
  tissue = tissue,
  comp = comp,
  pathways = pathways,
  celltypes = rev(celltypes),
  celltype_level = "l3",
  input_dir = "../../data/figure-inputs/",
  timepoint_1 = "EI",
  timepoint_2 = "ASCT1y",
  title = "BMMC - EI vs ASCT1y",
  panel_heights = c(1.5, 4),
  cap = 2,
  x_text_size = 14,
  base_size = 12,
  plot_width = 20,
  plot_height = 6
)

p
ggsave("../../data/figure-outputs/figure-4/fig_4d_p1.pdf", plot = p, width = 20, height = 6, dpi = 200, device = cairo_pdf)
```

```r
# Load bone marrow data files
bm_ei_v_1y <- read.csv('../../data/figure-inputs/l3-fgsea_all_results/fgsea_all_results_bmmc_EI_vs_ASCT1y.csv')

# Create list of datasets and comparison names
data_list <- list(bm_ei_v_1y)
bm_comparison_names <- c("EI vs ASCT1y")
```

```r
plot_width = 5
plot_height = 6
plot_dpi = 200

# faceted heatmap for bone marrow
p <- create_pathway_heatmap(
  data = bm_ei_v_1y,
  pathway_method = "manual",
  pathway_list = pathways,
  celltypes = rev(celltypes),
  max_label_length = 19,
  plot_width = plot_width,
  plot_height = plot_height,
  facet_ncol = 4
)

p
ggsave("../../data/figure-outputs/figure-4/fig_4d_p2.pdf", plot = p, width = plot_width, height = plot_height, dpi = plot_dpi, device = cairo_pdf)
```
