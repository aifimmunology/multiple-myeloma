```python
bmmc = sc.read_h5ad('../../data/figure-inputs/scrna_objects/final-bmmc-processed.h5ad', backed='r')
```

```python
l2_cmap_edited = {k: ('#BDBDBD' if "Plasma" in k else v) for k, v in l2_cmap_expanded.items()}
sc.pl.umap(
    bmmc,
    color=['aifi_plot_l2'],
    palette=l2_cmap_edited,
    size=2,
    title = '',
    legend_loc=None, 
    save='fig_2g.png'
)
```
