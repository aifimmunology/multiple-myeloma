```python
import scanpy as sc

# Set the output directory
sc.settings.figdir = '../../data/figure-outputs/figure-2'

# Set the visual parameters
sc.settings.set_figure_params(
    dpi=80,
    facecolor='white',
    frameon=False,
    fontsize=30,
    dpi_save=200,
    figsize=(6,6),
    format='png'
)
```

```python
adata = sc.read_h5ad('../../data/figure-inputs/scrna_objects/pretx-plasma-paga.h5ad', backed='r')
```

```python
genes = [
    "IL6ST",
    "IL2RG",
    "HGF",
    "MET",
    "VEGFB",
    "CCR10",
    "CCR1",
    "CCL3",
    "CCL5",
    "TNFRSF13B",
    "TNFRSF13C",
    "TNFRSF17"
]

sc.pl.umap(
    adata, color=genes, size=15, cmap="Reds", vmax="p99", ncols=6, use_raw=False, save='fig_2e.png'
)
```
