```r
# Load bone marrow data files
bm_ei_v_1y <- read.csv('../../data/figure-inputs/l3-fgsea_all_results/fgsea_all_results_bmmc_EI_vs_ASCT1y.csv')

# Create list of datasets and comparison names
data_list <- list(bm_ei_v_1y)
bm_comparison_names <- c("EI vs ASCT1y")
```

```r
pathways <- c(
  "NELFE ENCODE",
  "TNF-ALPHA SIGNALING VIA NF-KB",
  "STAT3 ENCODE",
  "SMC3 ENCODE",
  "NFE2L2 CHEA",
  "INTERFERON SIGNALING",
  "CYTOKINE SIGNALING IN IMMUNE SYSTEM"
)
```

```r
plot_width = 6
plot_height = 4
plot_dpi = 200

# faceted heatmap for bone marrow
p <- create_pathway_heatmap(
  data = bm_ei_v_1y,
  pathway_method = "manual",
  pathway_list = pathways,
  celltypes = rev(c("CD8 T EM1", "CD8 T EM2", "CD8 T Tissue Res")),
  max_label_length = 19,
  plot_width = plot_width,
  plot_height = plot_height,
  facet_ncol = 4
)

p
ggsave("../../data/figure-outputs/figure-5/fig_5i_p2.pdf", plot = p, width = plot_width, height = plot_height, dpi = plot_dpi, device = cairo_pdf)
```
