```python
import matplotlib.pyplot as plt
import pandas as pd
import scanpy as sc

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)

# Set the visual parameters
sc.settings.set_figure_params(dpi=200, facecolor="white", frameon=False)
```

## BMMC

```python
adata = sc.read_h5ad("../../data/figure-inputs/scrna_objects/final-bmmc-processed.h5ad")
```

```python
celltypes = [
    "CD56dim NK GZMK+",
    "NK Adaptive",
    "CD56dim NK GZMK-",
    "CD8 T EM2",
    "CD8 T EM1",
    "CD8 T Tissue Res"
]
```

```python
adata_sub = adata[(adata.obs["aifi_plot_l3"].isin(celltypes))].copy()

adata_sub.obs["aifi_plot_l3"] = pd.Categorical(
    adata_sub.obs["aifi_plot_l3"], categories=celltypes, ordered=True
)
```

```python
genes = [
    "XCL1",
    "KLRC1",
    "TRDC",
    "FGFBP2",
    "LGALS1",
    "SPON2",
    "KLRB1",
    "TYROBP",
    "FCER1G",
    "FCGR3A",
    "DUSP2",
    "IL7R",
    "CD69",
    "CD27",
    "GZMK",
    "GZMB",
    "NOSIP",
    "MX1",
    "OAS1",
    "CX3CR1",
    "TRGC2",
    "KLRG1",
    "KLRF1",
    "CD8B",
    "GZMH"
]
```

```python
plt.rcParams["pdf.fonttype"] = 42

sc.pl.dotplot(
    adata_sub,
    var_names=genes,
    groupby="aifi_plot_l3",
    dendrogram=False,
    swap_axes=False,
    show=False,
    figsize=(8, 2),
)

output_path = "../../data/figure-outputs/figure-5/fig_5a_p1.pdf"
plt.savefig(output_path, dpi=200, bbox_inches="tight")
```

## PBMC

```python
adata = sc.read_h5ad("../../data/figure-inputs/scrna_objects/final-pbmc-raw.h5ad")
```

```python
celltypes = [
    "CD56dim NK GZMK+",
    "NK Adaptive",
    "CD56dim NK GZMK-",
    "CD8 T EM GZMK+ CD27+",
    "CD8 T EM GZMK- CD27+",
    "CD8 T CM",
    "CD8 T Mem ISG+",
    "CD8 T EM KLRF1+ GZMB+",
    "CD8 T EM KLRF1- GZMB+",
]
```

```python
adata_sub = adata[(adata.obs["aifi_plot_l3"].isin(celltypes))].copy()

adata_sub.obs['aifi_plot_l3'] = pd.Categorical(
    adata_sub.obs['aifi_plot_l3'],
    categories=celltypes,
    ordered=True
)

sc.pp.normalize_total(adata_sub)
sc.pp.log1p(adata_sub)
```

```python
plt.rcParams["pdf.fonttype"] = 42

sc.pl.dotplot(
    adata_sub,
    var_names=genes,
    groupby="aifi_plot_l3",
    dendrogram=False,
    swap_axes=False,
    show=False,
    figsize=(8, 3),
)

output_path = "../../data/figure-outputs/figure-5/fig_5a_p2.pdf"
plt.savefig(output_path, dpi=200, bbox_inches="tight")
```
