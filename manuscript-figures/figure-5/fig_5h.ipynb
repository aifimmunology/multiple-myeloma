```r
plot_width = 12
plot_height = 4.5
plot_dpi = 200

p <- plot_volcano_publication(
  input_deg_res = "../../data/figure-inputs/l3-deseq2_results/deseq2_results_bmmc_EI_vs_ASCT1y.csv",
  selected_celltypes = c("CD8 T EM1", "CD8 T EM2", "CD8 T Tissue Res"),
  ncols = 3,
  timepoint_1 = "EI",
  timepoint_2 = "ASCT1y",
  title = 'BMMC'
)
options(repr.plot.width = plot_width, repr.plot.height = plot_height)
p

ggsave("../../data/figure-outputs/figure-5/fig_5h_p1.pdf", plot = p, width = plot_width, height = plot_height, dpi = plot_dpi, device = cairo_pdf)
```
