```r
genes = c('IFNG', 'TNF', 'TGFB1')

p1 <- plot_expression_violins(
  metadata_path = '../../data/figure-inputs/metadata/bmmc_sample_kit_metadata.csv',
  files_dir = '../../data/figure-inputs/bmmc_l3_lognorm_gexp_celltypes_per_samplekit',
  gene_list =  genes,
  celltypes = c('CD8 T EM1', 'CD8 T EM2', 'CD8 T Tissue Res'),
  visits = c('PreTx', 'EI', 'Healthy'),
  paired_visits = c('PreTx', 'EI'),
  unpaired_visit = 'Healthy',
  plot_width = 6,
  plot_height = 6,
  base_size = 10, 
  step.increase = 0.06,
  subtitle = '',
  ylabs = 'Pseudobulk Gene Expression',
  pval_colors = c('maroon', 'seagreen', 'gray30', 'gray50'),
  line_size = 0.6
)

p2 <- plot_expression_violins(
  metadata_path = '../../data/figure-inputs/metadata/pbmc_sample_kit_metadata.csv',
  files_dir = '../../data/figure-inputs/pbmc_l3_lognorm_gexp_celltypes_per_samplekit',
  gene_list =  genes,
  celltypes = c('CD8 T EM GZMK+ CD27+', 'CD8 T EM KLRF1+ GZMB+', 'CD8 T EM KLRF1- GZMB+'),
  visits = c('PreTx', 'EI', 'Healthy'),
  paired_visits = c('PreTx', 'EI'),
  unpaired_visit = 'Healthy',
  plot_width = 6,
  plot_height = 6,
  base_size = 10, 
  step.increase = 0.06,
  subtitle = '',
  pval_colors = c('maroon', 'seagreen', 'gray30', 'gray50'),
  line_size = 0.6
)

options(repr.plot.width = 10, repr.plot.height = 6)
p <- (p1 | p2) +
  plot_annotation(
    title = '',
    theme = theme(plot.title = element_text(face = 'bold', hjust = 0.5, size = 14))
  )

p
ggsave("../../data/figure-outputs/figure-5/fig_5g.pdf", plot = p, width = 10, height = 6, dpi = 200, device = cairo_pdf)
```
