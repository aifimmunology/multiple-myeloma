```r
tissue <- "bmmc"
comp <- "PreTx_vs_EI"

pathways <- c(
      "IL-17 SIGNALING PATHWAY",
      "TNF-ALPHA SIGNALING VIA NF-KB",
  "CYTOKINE SIGNALING IN IMMUNE SYSTEM",
  "MAPK SIGNALING PATHWAY"
)

celltypes <- c(
  "HSPC Stem", "HSPC Multi", "LMPP", "CLP",
  "Prog B Pre", "Prog B",
  "CMP Core", "CMP Gran",
  "MEP", "Pre Prog Ery", "Pre Mono Core",
  "Pre Mono Prolif",
  "Prog DC cDC", "Prog DC pDC"
)

p <- plot_pathway_genes_heatmap(
  tissue = tissue,
  comp = comp,
  pathways = pathways,
  celltypes = rev(celltypes),
  celltype_level = "l3",
  input_dir = "../../data/figure-inputs/",
  timepoint_1 = "PreTx",
  timepoint_2 = "EI",
  title = "BMMC - PreTx vs EI",
  panel_heights = c(1.5, 4),
  cap = 2,
  x_text_size = 14,
  base_size = 12,
  plot_width = 20,
  plot_height = 7
)

p
ggsave("../../data/figure-outputs/supplementary-4/sup_4f_p1.pdf", plot = p, width = 20, height = 7, dpi = 200, device = cairo_pdf)
```

```r
# Load bone marrow data files
data <- read.csv('../../data/figure-inputs/l3-fgsea_all_results/fgsea_all_results_bmmc_PreTx_vs_EI.csv')
```

```r
plot_width = 5
plot_height = 6
plot_dpi = 200

# faceted heatmap for bone marrow
p <- create_pathway_heatmap(
  data = data,
  pathway_method = "manual",
  pathway_list = pathways,
  celltypes = rev(celltypes),
  max_label_length = 19,
  plot_width = plot_width,
  plot_height = plot_height,
  facet_ncol = 4
)

p
ggsave("../../data/figure-outputs/supplementary-4/sup_4f_p2.pdf", plot = p, width = plot_width, height = plot_height, dpi = plot_dpi, device = cairo_pdf)
```
