```r
suppressPackageStartupMessages({require(data.table)
require(ggplot2)
require(ggpubr)
require(dplyr)
})
```

```r
# load CV data 
FH1 <- fread("../../data/figure-inputs/msd/flu_response_cv_classification.csv")
```

```r
options(repr.plot.width = 9, repr.plot.height = 7)

p <- ggplot(
  FH1,
  aes(
    y = FluCV,
    x = Flu_Season,
    group = Subject,
    label = Subject
  )
) +
  geom_point() +
  ggrepel::geom_text_repel(size=5) +
  facet_wrap(~Year, ncol = 2) +
  geom_hline(yintercept = 0.5, lty = 2, col = "green4", lwd = 2) +
  theme_linedraw(base_size = 12) +
  theme(
    text = element_text(size = 18),
    plot.title = element_text(face = "bold"),
    axis.text.x = element_text(angle = 45, hjust = 1),
    strip.background = element_rect(fill = "white", color = NA),
    strip.text = element_text(color = "black", size=18)
  )  +
  ylab("Flu Response Variation") +
  xlab("Flu Season")
p

ggsave("../../data/figure-outputs/supplementary-7/sup_7a.pdf", plot = p, width = 9, height = 7, dpi = 200, device = cairo_pdf)
```
