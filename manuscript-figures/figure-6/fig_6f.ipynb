```r
tissue <- "pbmc"
comp <- "ASCT1y_vs_Healthy"

pathways <- c(
  "EPSTEIN-BARR VIRUS INFECTION",
  "TNF-ALPHA SIGNALING VIA NF-KB",
  "INTERFERON ALPHA RESPONSE",
  "INTERFERON GAMMA RESPONSE"
)

celltypes <- c(
  "Mem B Core",
  "Naive B Core",
  "Naive B ISG+",
  "Trans B",
  "cDC1",
  "pDC"
)

p <- plot_pathway_genes_heatmap(
  tissue = tissue,
  comp = comp,
  pathways = pathways,
  celltypes = celltypes,
  celltype_level = "l3",
  input_dir = "../../data/figure-inputs/",
  timepoint_1 = "ASCT1y",
  timepoint_2 = "Healthy",
  title = "PBMC",
  panel_heights = c(3, 4),
  cap = 2,
  x_text_size = 14,
  base_size = 12,
  plot_width = 23,
  plot_height = 5,
  top_n_genes = 20
)

p
ggsave("../../data/figure-outputs/figure-6/fig_6f_p1.pdf", plot = p, width = 23, height = 5, dpi = 200, device = cairo_pdf)
```
