## Round 2 sub clustering

```python
adata_raw = sc.read_h5ad('../../../data/rna/bmmc-subsets/bmmc-mono_dc-processed-harmony.h5ad')
clusters = ['0', '13', '14']

mask = adata_raw.obs['leiden'].astype(str).isin(clusters)
adata = adata_raw[mask, :].copy()

adata = process_adata(adata, resolution=1.5, run_harmony=True, run_rank_genes=True)
adata.write('../../../data/rna/bmmc-subsets/bmmc-mono_dc-subcluster-processed-harmony.h5ad')

for cluster in clusters:
    mask = adata_raw.obs['leiden'].astype(str) == str(cluster)
    adata = adata_raw[mask, :]
    adata = process_adata(adata, resolution=1, run_harmony=True, run_rank_genes=True)
    adata.write(f'../../../data/rna/bmmc-subsets/bmmc-mono_dc-sub-{cluster}-processed-harmony.h5ad')
```

```python
adata_raw = sc.read_h5ad('../../../data/rna/pbmc-subsets/pbmc-mono_dc-processed-harmony.h5ad')
clusters = ['0', '1', '2', '3', '16']

mask = adata_raw.obs['leiden'].astype(str).isin(clusters)
adata = adata_raw[mask, :].copy()

adata = process_adata(adata, resolution=1.5, run_harmony=True, run_rank_genes=True)
adata.write('../../../data/rna/pbmc-subsets/pbmc-mono_dc-subcluster-processed-harmony.h5ad')

for cluster in clusters:
    mask = adata_raw.obs['leiden'].astype(str) == str(cluster)
    adata = adata_raw[mask, :]
    adata = process_adata(adata, resolution=1, run_harmony=True, run_rank_genes=True)
    adata.write(f'../../../data/rna/pbmc-subsets/pbmc-mono_dc-sub-{cluster}-processed-harmony.h5ad')
```
